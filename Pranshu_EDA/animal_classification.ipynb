# 1. Importing important libraires

```python
import pandas as pd
import re
import os
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
```

# 2. Filter for animal-related cases

```python
stop_words = set([
    'a', 'an', 'the', 'and', 'or', 'in', 'on', 'at', 'to', 'for',
    'with', 'is', 'are', 'was', 'were', 'by', 'of', 'from', 'that', 'this'
])

def clean_text(text):
    if pd.isnull(text): return ""
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    return " ".join([w for w in text.split() if w not in stop_words])
```

# 3. Define Stopwords and Cleaner 

```python
import os
import pandas as pd


data_folder = "."  # current folder

years = [21, 22, 23, 24, 25]
combined_dfs = []

for year in years:
    filename = os.path.join(data_folder, f"{year}.csv")
    if os.path.exists(filename):
        print(f" Found: {filename}")
        df = pd.read_csv(filename)
        df["year"] = 2000 + year
        combined_dfs.append(df)
    else:
        print(f" Missing file for year {year}: {filename}")

if combined_dfs:
    combined_df = pd.concat(combined_dfs, ignore_index=True)
    combined_df.to_csv("combined_data.csv", index=False)
    print(" Combined dataset saved as 'combined_data.csv' with shape:", combined_df.shape)
else:
    print("⚠️ No files were loaded. Check your working directory and file names.")
```

# 5. Filter Animal-Related Cases

```python
animal_df = combined_df[combined_df['case_title'].str.contains("animal", case=False, na=False)].copy()
print("Filtered animal-related records:", animal_df.shape)
```

# 6. Combine & Clean Text Fields

```python
animal_df["combined_text"] = animal_df["case_title"].fillna("") + " " + animal_df["closure_reason"].fillna("")
animal_df["cleaned_text"] = animal_df["combined_text"].apply(clean_text)
```

# 7.  TF-IDF + KMeans Clustering

```python
tfidf = TfidfVectorizer(max_features=500)
X = tfidf.fit_transform(animal_df["cleaned_text"])

kmeans = KMeans(n_clusters=5, random_state=42)
animal_df["animal_category"] = kmeans.fit_predict(X)
```

# 8. Inspect Cluster Samples

```python
for i in range(5):
    print(f"\n--- Cluster {i} ---")
    print(animal_df[animal_df["animal_category"] == i]["case_title"].sample(5, random_state=42).tolist())
```

# 9. Saving Output


```python
animal_df.to_csv("refined_animal_classification.csv", index=False)
print("Saved to refined_animal_classification.csv")
```
